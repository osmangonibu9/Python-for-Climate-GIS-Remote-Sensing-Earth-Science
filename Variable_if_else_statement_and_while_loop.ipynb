### If, elif, else statement

```python
a=False
if a:
    print('a is true')
if not a:
    print('a is false')
```

```python
a=bool(1)
b=bool(0)
if a:
    print('a is true')
    if b:
        print('a and b is True')
    if not b:
        print('a true but b is false')
elif not a:
    print('a is false')
    if b:
        print('a false and b is True')
    if not b:
        print('a and b is false')
```

```python
number=10
if number>10:
    print('The number is greater than 10')
elif number<10:
    print('The number is less than 10')
else:
    print('The number is equal')
```
